```python
# Dataset class with oversampling and augmentation
# Wait time is included to avoid Google Drive file access limit

import os
import time
import torch
from torch.utils.data import Dataset
from PIL import Image
from collections import Counter


class FruitClassificationDataset(Dataset):

    def __init__(self, main_dir, transform=None, balance=None):

        self.main_dir = main_dir
        self.transform = transform
        self.balance = balance

        self.classes = sorted(os.listdir(main_dir))

        self.quality_types = [
            'Good_Quality',
            'Bad_Quality',
            'Mixed_Quality'
        ]

        self.class_to_idx = {
            cls: idx for idx, cls in enumerate(self.classes)
        }

        self.all_imgs = [
            os.path.join(dp, f)
            for dp, dn, fn in os.walk(
                os.path.expanduser(main_dir)
            )
            for f in fn
            if f.endswith('.jpg') or f.endswith('.png')
        ]

        self.imgs_labelled = [
            (
                img,
                self.class_to_idx[
                    os.path.basename(
                        os.path.dirname(
                            os.path.dirname(img)
                        )
                    )
                ]
            )
            for img in self.all_imgs
        ]

        self.class_counts = Counter(
            img[1] for img in self.imgs_labelled
        )

        self.max_class_count = max(
            self.class_counts.values()
        )

        # Pre-calculate the list of samples for each class
        self.samples_by_class = [
            [
                img
                for img in self.imgs_labelled
                if img[1] == i
            ]
            for i in range(len(self.classes))
        ]

    def __len__(self):

        return len(self.classes) * self.max_class_count

    def wait_for_file(
        self,
        file_path,
        max_wait_time=120,
        delay=5
    ):

        total_wait_time = 0

        while total_wait_time < max_wait_time:

            if os.path.isfile(file_path):
                return True

            else:
                time.sleep(delay)
                total_wait_time += delay

        return False

    def __getitem__(self, index):

        class_index = index % len(self.classes)

        img_index = index // len(self.classes)

        class_samples = self.samples_by_class[class_index]

        img_index = img_index % len(class_samples)

        img_name, label = class_samples[img_index]

        # Wait for the image file to be accessible
        if not self.wait_for_file(img_name):
            raise FileNotFoundError(
                f"File not found after waiting: {img_name}"
            )

        # Load image
        image = Image.open(img_name).convert("RGB")

        # Apply oversampling augmentation
        if (
            self.balance
            and len(class_samples) < self.max_class_count
        ):
            image = self.balance(image)

        # Apply normal transformation
        if self.transform:
            image = self.transform(image)

        return image, label
```

```python
# Multi-label dataset class used for testing both models

import os
import time
from torch.utils.data import Dataset
from PIL import Image


class FruitDataset(Dataset):

    def __init__(self, main_dir, preprocess=None):

        self.main_dir = main_dir
        self.preprocess = preprocess

        # Fruit classes
        self.fruit_types = sorted([
            'Apple',
            'Banana',
            'Guava',
            'Lime',
            'Orange',
            'Pomegranate'
        ])

        self.fruit_to_idx = {
            fruit: idx
            for idx, fruit in enumerate(self.fruit_types)
        }

        # Quality classes
        self.quality_types = sorted([
            'Good_Quality',
            'Bad_Quality',
            'Mixed_Quality'
        ])

        self.quality_to_idx = {
            quality: idx
            for idx, quality in enumerate(self.quality_types)
        }

        # Find all images
        self.all_imgs = [
            os.path.join(dp, f)
            for dp, dn, fn in os.walk(
                os.path.expanduser(main_dir)
            )
            for f in fn
            if f.endswith('.jpg') or f.endswith('.png')
        ]

        # Assign both fruit and quality labels
        self.imgs_labelled = [
            (
                img,
                self.fruit_to_idx[
                    os.path.basename(
                        os.path.dirname(
                            os.path.dirname(img)
                        )
                    )
                ],
                self.quality_to_idx[
                    os.path.basename(
                        os.path.dirname(img)
                    )
                ]
            )
            for img in self.all_imgs
        ]

    def __len__(self):

        return len(self.imgs_labelled)

    def wait_for_file(
        self,
        file_path,
        max_wait_time=120,
        delay=5
    ):

        total_wait_time = 0

        while total_wait_time < max_wait_time:

            if os.path.isfile(file_path):
                return True

            else:
                time.sleep(delay)
                total_wait_time += delay

        return False

    def __getitem__(self, index):

        img_name, fruit_label, quality_label = (
            self.imgs_labelled[index]
        )

        # Wait for image file
        if not self.wait_for_file(img_name):

            raise FileNotFoundError(
                f"File not found after waiting: {img_name}"
            )

        # Load image
        image = Image.open(img_name).convert("RGB")

        # Apply preprocessing
        if self.preprocess:
            image = self.preprocess(image)

        return image, fruit_label, quality_label
```

```python
# Data Augmentation & Pre-processing

import torch
from torchvision.transforms import v2


# Define transformations for training set
transform = v2.Compose([
    v2.Resize(size=(224, 224), antialias=True),
    v2.RandomHorizontalFlip(),
    v2.RandomVerticalFlip(),
    v2.RandomRotation(degrees=45),
    v2.GaussianBlur(kernel_size=3),
    v2.ToImage(),
    v2.ToDtype(torch.uint8, scale=True),
    v2.ToDtype(torch.float32, scale=True),
    v2.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


# Define pre-processing for val and test sets
preprocess = v2.Compose([
    v2.Resize(size=(224, 224), antialias=True),
    v2.ToImage(),
    v2.ToDtype(torch.uint8, scale=True),
    v2.ToDtype(torch.float32, scale=True),
    v2.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


# Define data augmentations for balancing dataset
balance = v2.Compose([
    v2.AugMix(
        severity=3,
        mixture_width=3,
        chain_depth=-1,
        alpha=1.
    )
])
```

```python
# Dataloaders

import os
import sys
import pandas as pd
from PIL import Image
from torch.utils.data import Dataset, DataLoader

# ============================================================
# Settings
# ============================================================

batch_size = 8
num_workers = 2

# Diagnostic step: List files in Google Drive to find the folder name
mydrive_path = "/content/drive/MyDrive"
if os.path.exists(mydrive_path):
    drive_contents = os.listdir(mydrive_path)
    print("Files and folders currently visible in MyDrive:", drive_contents)
else:
    print("MyDrive path not found!")
    drive_contents = []

# Dynamically search for the correct project directory name using list of possibilities
possible_paths = [
    "/content/drive/MyDrive/Fruit_Classification_project",
    "/content/drive/MyDrive/Fruit_Classification_Project",
    "/content/drive/MyDrive/Fruit Classification project",
    "/content/drive/MyDrive/Fruit Classification Project"
]

PROJECT_PATH = None
for path in possible_paths:
    if os.path.exists(path):
        PROJECT_PATH = path
        break

# If no match in potential candidates, try to auto-detect any folder starting with 'Fruit' (excluding checkpoints folder)
if PROJECT_PATH is None and os.path.exists(mydrive_path):
    for name in drive_contents:
        if name.lower().startswith("fruit") and "checkpoint" not in name.lower():
            test_path = os.path.join(mydrive_path, name)
            if os.path.isdir(test_path):
                PROJECT_PATH = test_path
                break

# If still not found, check if splits directory exists inside any of the folders
if PROJECT_PATH is None and os.path.exists(mydrive_path):
    for name in drive_contents:
        test_path = os.path.join(mydrive_path, name)
        if os.path.isdir(test_path) and os.path.exists(os.path.join(test_path, "splits")):
            PROJECT_PATH = test_path
            break

if PROJECT_PATH is None:
    print("\n[WARNING] Could not find the main project folder containing 'splits/train.csv' in MyDrive.")
    print("Please make sure your project dataset folder (e.g. 'Fruit_Classification_project') is uploaded directly to your Google Drive root.")
    # Use the default placeholder to avoid breaking definition setup
    PROJECT_PATH = "/content/drive/MyDrive/Fruit_Classification_project"
else:
    print(f"\nSelected PROJECT_PATH: {PROJECT_PATH}")

SPLIT_PATH = os.path.join(PROJECT_PATH, "splits")

train_csv = os.path.join(SPLIT_PATH, "train.csv")
val_csv = os.path.join(SPLIT_PATH, "val.csv")
test_csv = os.path.join(SPLIT_PATH, "test.csv")

# Check if train.csv is missing to stop execution early with a clean instructions message
if not os.path.exists(train_csv):
    print(f"\n[ERROR] The file '{train_csv}' was not found.")
    print("Please verify that the dataset files are fully synced to your Google Drive and try running this cell again.")
else:
    # ============================================================
    # CSV Dataset
    # ============================================================

    class CSVFruitDataset(Dataset):

        def __init__(
            self,
            csv_path,
            transform=None,
            balance=None,
            target="fruit"
        ):

            self.dataframe = pd.read_csv(csv_path)

            # Clean up image paths to point to the dynamically detected PROJECT_PATH
            if "image_path" in self.dataframe.columns:
                self.dataframe["image_path"] = self.dataframe["image_path"].str.replace(
                    "/content/drive/MyDrive/Fruit Classification project/",
                    PROJECT_PATH + "/",
                    regex=False
                ).str.replace(
                    "/content/drive/MyDrive/Fruit Classification Project/",
                    PROJECT_PATH + "/",
                    regex=False
                ).str.replace(
                    "/content/drive/MyDrive/Fruit_Classification_Project/",
                    PROJECT_PATH + "/",
                    regex=False
                ).str.replace(
                    "/content/drive/MyDrive/Fruit_Classification_project/",
                    PROJECT_PATH + "/",
                    regex=False
                )

            self.transform = transform
            self.balance = balance
            self.target = target

            # Fruit classes
            self.fruit_types = [
                "Apple",
                "Banana",
                "Guava",
                "Lime",
                "Orange",
                "Pomegranate"
            ]

            self.fruit_to_idx = {
                fruit: idx
                for idx, fruit in enumerate(self.fruit_types)
            }

            # Quality classes
            self.quality_types = [
                "Good",
                "Bad",
                "Mixed"
            ]

            self.quality_to_idx = {
                quality: idx
                for idx, quality in enumerate(self.quality_types)
            }

        def __len__(self):

            return len(self.dataframe)

        def __getitem__(self, index):

            row = self.dataframe.iloc[index]

            image_path = row["image_path"]

            image = Image.open(image_path).convert("RGB")

            # Apply AugMix for balancing
            if self.balance is not None:

                if self.target == "fruit":

                    fruit_count = (
                        self.dataframe["fruit"]
                        .value_counts()
                        .get(row["fruit"], 0)
                    )

                    max_fruit_count = (
                        self.dataframe["fruit"]
                        .value_counts()
                        .max()
                    )

                    if fruit_count < max_fruit_count:
                        image = self.balance(image)

                elif self.target == "quality":

                    quality_count = (
                        self.dataframe["quality"]
                        .value_counts()
                        .get(row["quality"], 0)
                    )

                    max_quality_count = (
                        self.dataframe["quality"]
                        .value_counts()
                        .max()
                    )

                    if quality_count < max_quality_count:
                        image = self.balance(image)

            # Apply normal transformation
            if self.transform is not None:

                image = self.transform(image)

            # Fruit classification
            if self.target == "fruit":

                label = self.fruit_to_idx[row["fruit"]]

            # Quality classification
            elif self.target == "quality":

                label = self.quality_to_idx[row["quality"]]

            return image, label

    # ============================================================
    # Fruit Classification Datasets
    # ============================================================

    f_train_dataset = CSVFruitDataset(
        train_csv,
        transform=transform,
        balance=balance,
        target="fruit"
    )

    f_val_dataset = CSVFruitDataset(
        val_csv,
        transform=preprocess,
        target="fruit"
    )

    f_test_dataset = CSVFruitDataset(
        test_csv,
        transform=preprocess,
        target="fruit"
    )

    # ============================================================
    # Fruit Classification DataLoaders
    # ============================================================

    f_train_loader = DataLoader(
        f_train_dataset,
        batch_size=batch_size,
        shuffle=True,
        num_workers=num_workers
    )

    f_val_loader = DataLoader(
        f_val_dataset,
        batch_size=batch_size,
        shuffle=False,
        num_workers=num_workers
    )

    f_test_loader = DataLoader(
        f_test_dataset,
        batch_size=batch_size,
        shuffle=False,
        num_workers=num_workers
    )

    # ============================================================
    # Quality Classification Datasets
    # ============================================================

    q_train_dataset = CSVFruitDataset(
        train_csv,
        transform=transform,
        balance=balance,
        target="quality"
    )

    q_val_dataset = CSVFruitDataset(
        val_csv,
        transform=preprocess,
        target="quality"
    )

    q_test_dataset = CSVFruitDataset(
        test_csv,
        transform=preprocess,
        target="quality"
    )

    # ============================================================
    # Quality Classification DataLoaders
    # ============================================================

    q_train_loader = DataLoader(
        q_train_dataset,
        batch_size=batch_size,
        shuffle=True,
        num_workers=num_workers
    )

    q_val_loader = DataLoader(
        q_val_dataset,
        batch_size=batch_size,
        shuffle=False,
        num_workers=num_workers
    )

    q_test_loader = DataLoader(
        q_test_dataset,
        batch_size=batch_size,
        shuffle=False,
        num_workers=num_workers
    )

    # ============================================================
    # Display Dataset Information
    # ============================================================

    print("Dataloaders created successfully.")
    print()

    print("Fruit Classification:")
    print("Train samples:", len(f_train_dataset))
    print("Validation samples:", len(f_val_dataset))
    print("Test samples:", len(f_test_dataset))

    print()

    print("Quality Classification:")
    print("Train samples:", len(q_train_dataset))
    print("Validation samples:", len(q_val_dataset))
    print("Test samples:", len(q_test_dataset))

    print()

    print("Batch size:", batch_size)
    print("Workers:", num_workers)

    print()

    print("Fruit train batches:", len(f_train_loader))
    print("Fruit validation batches:", len(f_val_loader))
    print("Fruit test batches:", len(f_test_loader))

    print()

    print("Quality train batches:", len(q_train_loader))
    print("Quality validation batches:", len(q_val_loader))
    print("Quality test batches:", len(q_test_loader))
```

```python
# Fruit Type Classification Model

import os
import copy
import glob
import torch
import torch.optim as optim
from torch import nn
from torch.optim.lr_scheduler import ReduceLROnPlateau
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
from torch.utils.tensorboard import SummaryWriter
from torchvision.models import efficientnet_v2_s
from torchvision.transforms import v2
from tqdm.auto import tqdm
import numpy as np

# ============================================================
# Guard Checks: Ensure Dataloaders are initialized first
# ============================================================
required_vars = [
    'f_train_dataset', 'f_val_dataset', 'f_test_dataset',
    'q_train_dataset', 'q_val_dataset', 'q_test_dataset',
    'f_train_loader', 'f_val_loader'
]
missing_vars = [var for var in required_vars if var not in globals()]

if missing_vars:
    print("[ERROR] Cannot start training because the required datasets/dataloaders are not initialized.")
    print("Please make sure you have:")
    print(" 1. Uploaded your dataset folder (with 'splits/train.csv') directly to Google Drive.")
    print(" 2. Successfully executed the Dataloaders cell (cell_id: Yuiyd1_-8-ph).")
else:
    # ============================================================
    # Load EfficientNetV2-S
    # ============================================================

    model = efficientnet_v2_s(
        weights='IMAGENET1K_V1',
        progress=True
    )

    # ============================================================
    # Number of classes
    # ============================================================

    num_classes = 6

    # ============================================================
    # CutMix and MixUp
    # ============================================================

    cutmix = v2.CutMix(
        num_classes=num_classes
    )

    mixup = v2.MixUp(
        num_classes=num_classes
    )

    cutmix_or_mixup = v2.RandomChoice(
        [cutmix, mixup]
    )

    # ============================================================
    # Modify the final classification layer
    # ============================================================

    num_features = model.classifier[1].in_features

    model.classifier[1] = nn.Linear(
        num_features,
        num_classes
    )

    # ============================================================
    # Loss function
    # ============================================================

    criterion = nn.CrossEntropyLoss()

    # ============================================================
    # Optimizer
    # Learning Rate = 0.01
    # ============================================================

    optimizer = optim.AdamW(
        model.parameters(),
        lr=0.01,
        weight_decay=1e-2
    )

    # ============================================================
    # Learning-rate scheduler
    # ============================================================

    scheduler = ReduceLROnPlateau(
        optimizer,
        mode='min',
        factor=0.1,
        patience=3
    )

    # ============================================================
    # Device
    # ============================================================

    device = torch.device(
        "cuda" if torch.cuda.is_available() else "cpu"
    )

    print("Using device:", device)

    model = model.to(device)

    # ============================================================
    # Checkpoint directory
    # ============================================================

    checkpoint_dir = "/content/drive/MyDrive/fruitcheckpoints"

    os.makedirs(
        checkpoint_dir,
        exist_ok=True
    )

    print("Checkpoint directory:")
    print(checkpoint_dir)

    # ============================================================
    # Path cleaning for all datasets
    # ============================================================

    def clean_dataset_paths(dataset):
        if hasattr(dataset, 'dataframe'):
            dataset.dataframe['image_path'] = dataset.dataframe['image_path'].str.replace(
                '/content/drive/MyDrive/Fruit Classification project/',
                '/content/drive/MyDrive/Fruit_Classification_project/',
                regex=False
            ).str.replace(
                '/content/drive/MyDrive/Fruit_Classification_Project/',
                '/content/drive/MyDrive/Fruit_Classification_project/',
                regex=False
            )

    clean_dataset_paths(f_train_dataset)
    clean_dataset_paths(f_val_dataset)
    clean_dataset_paths(f_test_dataset)
    clean_dataset_paths(q_train_dataset)
    clean_dataset_paths(q_val_dataset)
    clean_dataset_paths(q_test_dataset)

    # ============================================================
    # Training settings
    # ============================================================

    num_epochs = 30

    start_epoch = 0

    best_val_accuracy = -np.inf

    best_model_state = copy.deepcopy(
        model.state_dict()
    )

    # ============================================================
    # Automatically find the latest checkpoint
    # ============================================================

    checkpoint_files = glob.glob(
        os.path.join(
            checkpoint_dir,
            "checkpoint_*.pth"
        )
    )

    if checkpoint_files:
        # Sort checkpoint files according to epoch number
        checkpoint_files.sort(
            key=lambda x: int(
                os.path.basename(x)
                .split("_")[-1]
                .split(".")[0]
            )
        )

        latest_checkpoint = checkpoint_files[-1]

        print("\nCheckpoint found:")
        print(latest_checkpoint)

        print("\nLoading checkpoint...")

        checkpoint = torch.load(
            latest_checkpoint,
            map_location=device
        )

        model.load_state_dict(
            checkpoint["model_state_dict"]
        )

        optimizer.load_state_dict(
            checkpoint["optimizer_state_dict"]
        )

        scheduler.load_state_dict(
            checkpoint["scheduler_state_dict"]
        )

        start_epoch = checkpoint["epoch"] + 1

        if "best_val_accuracy" in checkpoint:
            best_val_accuracy = checkpoint[
                "best_val_accuracy"
            ]

        best_model_state = copy.deepcopy(
            model.state_dict()
        )

        print("Checkpoint loaded successfully.")
        print(f"Resuming from Epoch {start_epoch + 1}")
    else:
        print("\nNo checkpoint found.")
        print("Starting training from Epoch 1.")

    # ============================================================
    # Training
    # ============================================================

    for epoch in range(
        start_epoch,
        num_epochs
    ):
        # Training mode
        model.train()
        running_loss = 0.0
        all_train_preds = []
        all_train_labels = []

        # Training progress bar
        train_progress = tqdm(
            enumerate(f_train_loader),
            total=len(f_train_loader),
            desc=f"Epoch {epoch + 1}/{num_epochs} - Training",
            unit="batch"
        )

        # Training batches
        for i, data in train_progress:
            inputs = data[0].to(device)
            labels = data[1].to(device)

            # Apply CutMix or MixUp
            inputs, labels = cutmix_or_mixup(
                inputs,
                labels
            )

            # Zero gradients
            optimizer.zero_grad()

            # Forward pass
            outputs = model(inputs)

            # Calculate loss
            loss = criterion(
                outputs,
                labels
            )

            # Backpropagation
            loss.backward()

            # Update weights
            optimizer.step()

            # Convert soft labels to hard labels
            hard_labels = torch.argmax(
                labels,
                dim=1
            )
            _, preds = torch.max(
                outputs,
                1
            )

            all_train_preds.extend(
                preds.detach().cpu().numpy()
            )
            all_train_labels.extend(
                hard_labels.detach().cpu().numpy()
            )

            # Running loss
            running_loss += loss.item()

            # Calculate current percentage
            progress_percent = (
                (i + 1) /
                len(f_train_loader)
            ) * 100

            # Update progress bar
            train_progress.set_postfix(
                loss=f"{loss.item():.4f}",
                progress=f"{progress_percent:.1f}%"
            )

        # Average training loss
        running_loss = (
            running_loss /
            len(f_train_loader)
        )

        # Training accuracy
        train_accuracy = accuracy_score(
            all_train_labels,
            all_train_preds
        )

        print(
            f"\nEpoch {epoch + 1} "
            f"loss: {running_loss:.3f}, "
            f"training accuracy: {train_accuracy:.3f}"
        )

        # TensorBoard
        writer.add_scalar(
            'Training Loss',
            running_loss,
            epoch
        )
        writer.add_scalar(
            'Training Accuracy',
            train_accuracy,
            epoch
        )

        # ========================================================
        # Validation
        # ========================================================

        model.eval()
        val_loss = 0.0
        all_preds = []
        all_labels = []

        # Validation progress bar
        val_progress = tqdm(
            enumerate(f_val_loader),
            total=len(f_val_loader),
            desc=f"Epoch {epoch + 1}/{num_epochs} - Validation",
            unit="batch"
        )

        # Validation batches
        with torch.no_grad():
            for i, val_data in val_progress:
                val_inputs = val_data[0].to(device)
                val_labels = val_data[1].to(device)

                # Forward pass
                val_outputs = model(
                    val_inputs
                )

                # Validation loss
                batch_val_loss = criterion(
                    val_outputs,
                    val_labels
                ).item()

                val_loss += batch_val_loss

                # Predictions
                _, preds = torch.max(
                    val_outputs,
                    1
                )

                all_preds.extend(
                    preds.cpu().numpy()
                )
                all_labels.extend(
                    val_labels.cpu().numpy()
                )

                # Validation percentage
                progress_percent = (
                    (i + 1) /
                    len(f_val_loader)
                ) * 100

                val_progress.set_postfix(
                    loss=f"{batch_val_loss:.4f}",
                    progress=f"{progress_percent:.1f}%"
                )

        # Average validation loss
        val_loss = (
            val_loss /
            len(f_val_loader)
        )

        # Validation accuracy
        val_accuracy = accuracy_score(
            all_labels,
            all_preds
        )

        print(
            f"\nEpoch {epoch + 1} "
            f"validation loss: {val_loss:.3f}, "
            f"validation accuracy: {val_accuracy:.3f}"
        )

        # TensorBoard validation metrics
        writer.add_scalar(
            'Validation Loss',
            val_loss,
            epoch
        )
        writer.add_scalar(
            'Validation Accuracy',
            val_accuracy,
            epoch
        )

        # ========================================================
        # Save checkpoint
        # ========================================================

        checkpoint_path = os.path.join(
            checkpoint_dir,
            f"checkpoint_{epoch}.pth"
        )

        torch.save(
            {
                'epoch': epoch,
                'model_state_dict': model.state_dict(),
                'optimizer_state_dict': optimizer.state_dict(),
                'scheduler_state_dict': scheduler.state_dict(),
                'loss': val_loss,
                'best_val_accuracy': best_val_accuracy
            },
            checkpoint_path
        )

        print(f"Checkpoint saved: {checkpoint_path}")

        # ========================================================
        # Check best model
        # ========================================================

        if val_accuracy > best_val_accuracy:
            best_val_accuracy = val_accuracy
            best_model_state = copy.deepcopy(
                model.state_dict()
            )

            # Save best model separately
            best_model_path = os.path.join(
                checkpoint_dir,
                "best_model.pth"
            )

            torch.save(
                {
                    'epoch': epoch,
                    'model_state_dict': model.state_dict(),
                    'optimizer_state_dict': optimizer.state_dict(),
                    'scheduler_state_dict': scheduler.state_dict(),
                    'best_val_accuracy': best_val_accuracy
                },
                best_model_path
            )

            print(f"Best model saved. Validation accuracy: {best_val_accuracy:.4f}")

        # ========================================================
        # Scheduler
        # ========================================================

        scheduler.step(
            val_loss
        )

        # Current learning rate
        current_lr = optimizer.param_groups[0]["lr"]
        print(f"Current learning rate: {current_lr}")

    print("\nFinished Training")
    print(f"Best Validation Accuracy: {best_val_accuracy:.4f}")
```

```python
# ============================================================
# QUALITY CLASSIFICATION MODEL
# Base-paper code + progress percentage + checkpoint resume
# ============================================================

import os
import torch
import torch.optim as optim
from torch import nn
from torch.optim.lr_scheduler import ReduceLROnPlateau
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
from torch.utils.tensorboard import SummaryWriter
import numpy as np
from torchvision.models import efficientnet_v2_s
from torchvision.transforms import v2
from tqdm.auto import tqdm


# ============================================================
# 1. Define device
# ============================================================

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Using device:", device)


# ============================================================
# 2. Define number of classes
# ============================================================

num_fruit_classes = 6
num_quality_classes = 3


# ============================================================
# 3. Create EfficientNetV2-S
# ============================================================

model = efficientnet_v2_s(
    weights='IMAGENET1K_V1',
    progress=True
)


# ============================================================
# 4. Change classifier to 6 fruit classes
# ============================================================

# 🔧 Necessary modification:
# MFC.pth contains weights for 6 fruit classes.
# The default EfficientNetV2-S has 1000 ImageNet classes.
#
# Therefore, the classifier must first be changed from:
# 1000 classes -> 6 fruit classes.

num_features = model.classifier[1].in_features

model.classifier[1] = nn.Linear(
    num_features,
    num_fruit_classes
)


# ============================================================
# 5. Load the saved fruit classification model
# ============================================================

fruit_model_path = (
    "/content/drive/MyDrive/MFC/MFC.pth"
)

fruit_checkpoint = torch.load(
    fruit_model_path,
    map_location=device
)

print("Fruit classification checkpoint loaded.")

print("Checkpoint keys:")
print(fruit_checkpoint.keys())


# ============================================================
# 6. Load fruit classification weights
# ============================================================

model.load_state_dict(
    fruit_checkpoint["model_state_dict"]
)

print(
    "Fruit classification weights loaded successfully."
)


# ============================================================
# 7. Define CutMix and MixUp
# ============================================================

cutmix = v2.CutMix(
    num_classes=num_quality_classes
)

mixup = v2.MixUp(
    num_classes=num_quality_classes
)

cutmix_or_mixup = v2.RandomChoice([
    cutmix,
    mixup
])


# ============================================================
# 8. Change classifier to 3 quality classes
# ============================================================

# Now transfer the learned fruit features to the
# quality classification task.

num_features = model.classifier[1].in_features

model.classifier[1] = nn.Linear(
    num_features,
    num_quality_classes
)

print(
    "Final classifier changed to",
    num_quality_classes,
    "quality classes."
)


# ============================================================
# 9. Move model to device
# ============================================================

model = model.to(device)


# ============================================================
# 10. Define loss function
# ============================================================

criterion = nn.CrossEntropyLoss()


# ============================================================
# 11. Define optimizer
# ============================================================

optimizer = optim.AdamW(
    model.parameters(),
    lr=0.01,
    weight_decay=1e-2
)


# ============================================================
# 12. Define learning-rate scheduler
# ============================================================

# 🔧 Necessary modification:
# Current PyTorch does not support verbose=True.

scheduler = ReduceLROnPlateau(
    optimizer,
    mode='min',
    factor=0.1,
    patience=3
)


# ============================================================
# 13. Create quality checkpoint directory
# ============================================================

checkpoint_dir = (
    "/content/drive/MyDrive/qualitycheckpoints"
)

os.makedirs(
    checkpoint_dir,
    exist_ok=True
)

print(
    "Checkpoint directory:",
    checkpoint_dir
)


# ============================================================
# 14. TensorBoard writer
# ============================================================

log_dir = (
    "/content/drive/MyDrive/log/quality_classification"
)

os.makedirs(
    log_dir,
    exist_ok=True
)

writer = SummaryWriter(
    log_dir=log_dir
)


# ============================================================
# 15. Training configuration
# ============================================================

num_epochs = 30

start_epoch = 0

best_val_accuracy = -np.inf


# ============================================================
# 16. Find latest quality checkpoint
# ============================================================

checkpoint_files = []

for file in os.listdir(checkpoint_dir):

    if (
        file.startswith("checkpoint_")
        and file.endswith(".pth")
    ):

        checkpoint_files.append(file)


if checkpoint_files:

    checkpoint_files.sort(
        key=lambda x: int(
            x.replace("checkpoint_", "")
             .replace(".pth", "")
        )
    )

    latest_checkpoint = checkpoint_files[-1]

    latest_checkpoint_path = os.path.join(
        checkpoint_dir,
        latest_checkpoint
    )

    print()
    print(
        "Latest quality checkpoint found:"
    )

    print(
        latest_checkpoint_path
    )


    # --------------------------------------------------------
    # Load quality checkpoint
    # --------------------------------------------------------

    resume_checkpoint = torch.load(
        latest_checkpoint_path,
        map_location=device
    )


    # --------------------------------------------------------
    # Restore model
    # --------------------------------------------------------

    model.load_state_dict(
        resume_checkpoint["model_state_dict"]
    )


    # --------------------------------------------------------
    # Restore optimizer
    # --------------------------------------------------------

    optimizer.load_state_dict(
        resume_checkpoint["optimizer_state_dict"]
    )


    # --------------------------------------------------------
    # Restore scheduler
    # --------------------------------------------------------

    if "scheduler_state_dict" in resume_checkpoint:

        scheduler.load_state_dict(
            resume_checkpoint["scheduler_state_dict"]
        )


    # --------------------------------------------------------
    # Restore epoch
    # --------------------------------------------------------

    start_epoch = (
        resume_checkpoint["epoch"] + 1
    )


    # --------------------------------------------------------
    # Restore best accuracy
    # --------------------------------------------------------

    if "best_val_accuracy" in resume_checkpoint:

        best_val_accuracy = (
            resume_checkpoint[
                "best_val_accuracy"
            ]
        )


    print(
        f"Resuming from epoch "
        f"{start_epoch + 1}"
    )

else:

    print()
    print(
        "No previous quality checkpoint found."
    )

    print(
        "Starting quality training from epoch 1."
    )


# ============================================================
# 17. Training loop
# ============================================================

for epoch in range(
    start_epoch,
    num_epochs
):

    # ========================================================
    # Training
    # ========================================================

    model.train()

    running_loss = 0.0

    all_train_preds = []

    all_train_labels = []


    print()
    print(
        f"Epoch {epoch + 1}/{num_epochs} - Training"
    )


    train_progress = tqdm(
        q_train_loader,
        total=len(q_train_loader),
        desc=(
            f"Epoch {epoch + 1}/{num_epochs} "
            f"- Training"
        ),
        unit="batch"
    )


    for i, data in enumerate(
        train_progress,
        0
    ):

        # ----------------------------------------------------
        # Get inputs and labels
        # ----------------------------------------------------

        inputs = data[0].to(device)

        labels = data[1].to(device)


        # ----------------------------------------------------
        # Apply CutMix or MixUp
        # ----------------------------------------------------

        inputs, labels = cutmix_or_mixup(
            inputs,
            labels
        )


        # ----------------------------------------------------
        # Zero gradients
        # ----------------------------------------------------

        optimizer.zero_grad()


        # ----------------------------------------------------
        # Forward pass
        # ----------------------------------------------------

        outputs = model(inputs)


        # ----------------------------------------------------
        # Calculate loss
        # ----------------------------------------------------

        loss = criterion(
            outputs,
            labels
        )


        # ----------------------------------------------------
        # Backward pass
        # ----------------------------------------------------

        loss.backward()


        # ----------------------------------------------------
        # Update model
        # ----------------------------------------------------

        optimizer.step()


        # ----------------------------------------------------
        # Convert soft labels to hard labels
        # ----------------------------------------------------

        hard_labels = torch.argmax(
            labels,
            dim=1
        )


        # ----------------------------------------------------
        # Predictions
        # ----------------------------------------------------

        _, preds = torch.max(
            outputs,
            1
        )


        # ----------------------------------------------------
        # Store predictions
        # ----------------------------------------------------

        all_train_preds.extend(
            preds.detach().cpu().numpy()
        )

        all_train_labels.extend(
            hard_labels.detach().cpu().numpy()
        )


        # ----------------------------------------------------
        # Running loss
        # ----------------------------------------------------

        running_loss += loss.item()


        # ----------------------------------------------------
        # Progress percentage
        # ----------------------------------------------------

        progress_percent = (
            (i + 1)
            / len(q_train_loader)
        ) * 100


        train_progress.set_postfix(
            loss=f"{loss.item():.4f}",
            progress=f"{progress_percent:.1f}%"
        )


    # ========================================================
    # Training metrics
    # ========================================================

    running_loss = (
        running_loss
        / len(q_train_loader)
    )

    train_accuracy = accuracy_score(
        all_train_labels,
        all_train_preds
    )


    print(
        f"Epoch {epoch + 1} loss: "
        f"{running_loss:.3f}, "
        f"training accuracy: "
        f"{train_accuracy:.3f}"
    )


    # ========================================================
    # TensorBoard
    # ========================================================

    writer.add_scalar(
        "Training Loss",
        running_loss,
        epoch
    )

    writer.add_scalar(
        "Training Accuracy",
        train_accuracy,
        epoch
    )


    # ========================================================
    # Validation
    # ========================================================

    model.eval()

    val_loss = 0.0

    all_preds = []

    all_labels = []


    print()
    print(
        f"Epoch {epoch + 1}/{num_epochs} - Validation"
    )


    val_progress = tqdm(
        q_val_loader,
        total=len(q_val_loader),
        desc=(
            f"Epoch {epoch + 1}/{num_epochs} "
            f"- Validation"
        ),
        unit="batch"
    )


    with torch.no_grad():

        for i, val_data in enumerate(
            val_progress,
            0
        ):

            val_inputs = (
                val_data[0].to(device)
            )

            val_labels = (
                val_data[1].to(device)
            )


            # ------------------------------------------------
            # Forward pass
            # ------------------------------------------------

            val_outputs = model(
                val_inputs
            )


            # ------------------------------------------------
            # Validation loss
            # ------------------------------------------------

            batch_val_loss = criterion(
                val_outputs,
                val_labels
            )


            val_loss += (
                batch_val_loss.item()
            )


            # ------------------------------------------------
            # Predictions
            # ------------------------------------------------

            _, preds = torch.max(
                val_outputs,
                1
            )


            all_preds.extend(
                preds.cpu().numpy()
            )

            all_labels.extend(
                val_labels.cpu().numpy()
            )


            # ------------------------------------------------
            # Progress percentage
            # ------------------------------------------------

            progress_percent = (
                (i + 1)
                / len(q_val_loader)
            ) * 100


            val_progress.set_postfix(
                loss=f"{batch_val_loss.item():.4f}",
                progress=f"{progress_percent:.1f}%"
            )


    # ========================================================
    # Validation metrics
    # ========================================================

    val_loss = (
        val_loss
        / len(q_val_loader)
    )

    val_accuracy = accuracy_score(
        all_labels,
        all_preds
    )


    print(
        f"Epoch {epoch + 1} validation loss: "
        f"{val_loss:.3f}, "
        f"validation accuracy: "
        f"{val_accuracy:.3f}"
    )


    # ========================================================
    # TensorBoard validation metrics
    # ========================================================

    writer.add_scalar(
        "Validation Loss",
        val_loss,
        epoch
    )

    writer.add_scalar(
        "Validation Accuracy",
        val_accuracy,
        epoch
    )


    # ========================================================
    # Save best model
    # ========================================================

    if val_accuracy > best_val_accuracy:

        best_val_accuracy = val_accuracy

        best_model_path = os.path.join(
            checkpoint_dir,
            "best_model.pth"
        )


        torch.save(
            {
                "epoch": epoch,
                "model_state_dict": model.state_dict(),
                "optimizer_state_dict": optimizer.state_dict(),
                "scheduler_state_dict": scheduler.state_dict(),
                "best_val_accuracy": best_val_accuracy
            },
            best_model_path
        )


        print(
            f"Best model saved. "
            f"Validation accuracy: "
            f"{best_val_accuracy:.4f}"
        )


    # ========================================================
    # Save checkpoint every epoch
    # ========================================================

    checkpoint_path = os.path.join(
        checkpoint_dir,
        f"checkpoint_{epoch}.pth"
    )


    torch.save(
        {
            "epoch": epoch,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "scheduler_state_dict": scheduler.state_dict(),
            "loss": val_loss,
            "best_val_accuracy": best_val_accuracy
        },
        checkpoint_path
    )


    print(
        f"Checkpoint saved: "
        f"{checkpoint_path}"
    )


    # ========================================================
    # Scheduler
    # ========================================================

    scheduler.step(
        val_loss
    )


    # ========================================================
    # Current learning rate
    # ========================================================

    current_lr = (
        optimizer.param_groups[0]["lr"]
    )

    print(
        f"Current learning rate: "
        f"{current_lr}"
    )


# ============================================================
# 18. Finished
# ============================================================

writer.close()

print()
print("==============================================")
print("Finished Quality Classification Training")
print("==============================================")

print(
    f"Best Validation Accuracy: "
    f"{best_val_accuracy:.4f}"
)
```

```python
# ============================================================
# Cell 22 - CBAM Module
# ============================================================

import torch
import torch.nn as nn


# ============================================================
# Channel Attention
# ============================================================

class ChannelAttention(nn.Module):

    def __init__(self, channels, reduction=16):

        super().__init__()

        hidden_channels = max(
            channels // reduction,
            1
        )

        self.avg_pool = nn.AdaptiveAvgPool2d(1)

        self.max_pool = nn.AdaptiveMaxPool2d(1)

        self.mlp = nn.Sequential(

            nn.Conv2d(
                channels,
                hidden_channels,
                kernel_size=1,
                bias=False
            ),

            nn.ReLU(),

            nn.Conv2d(
                hidden_channels,
                channels,
                kernel_size=1,
                bias=False
            )
        )

        self.sigmoid = nn.Sigmoid()


    def forward(self, x):

        avg_out = self.mlp(
            self.avg_pool(x)
        )

        max_out = self.mlp(
            self.max_pool(x)
        )

        attention = avg_out + max_out

        attention = self.sigmoid(
            attention
        )

        return x * attention


# ============================================================
# Spatial Attention
# ============================================================

class SpatialAttention(nn.Module):

    def __init__(self, kernel_size=7):

        super().__init__()

        self.conv = nn.Conv2d(
            2,
            1,
            kernel_size=kernel_size,
            padding=kernel_size // 2,
            bias=False
        )

        self.sigmoid = nn.Sigmoid()


    def forward(self, x):

        avg_out = torch.mean(
            x,
            dim=1,
            keepdim=True
        )

        max_out = torch.max(
            x,
            dim=1,
            keepdim=True
        )[0]

        attention = torch.cat(
            [avg_out, max_out],
            dim=1
        )

        attention = self.conv(
            attention
        )

        attention = self.sigmoid(
            attention
        )

        return x * attention


# ============================================================
# CBAM
# ============================================================

class CBAM(nn.Module):

    def __init__(
        self,
        channels,
        reduction=16,
        kernel_size=7
    ):

        super().__init__()

        self.channel_attention = ChannelAttention(
            channels=channels,
            reduction=reduction
        )

        self.spatial_attention = SpatialAttention(
            kernel_size=kernel_size
        )


    def forward(self, x):

        x = self.channel_attention(x)

        x = self.spatial_attention(x)

        return x


print("CBAM module created successfully.")
```

```python
# ============================================================
# Cell 23 - EfficientNetV2-S + CBAM + Multi-Task Learning
# ============================================================

from torchvision.models import efficientnet_v2_s


# ============================================================
# EfficientNetV2-S + CBAM + MTL Model
# ============================================================

class EfficientNetV2S_CBAM_MTL(nn.Module):

    def __init__(
        self,
        num_fruit_classes=6,
        num_quality_classes=3
    ):

        super().__init__()


        # ----------------------------------------------------
        # Load ImageNet pretrained EfficientNetV2-S
        # ----------------------------------------------------

        self.backbone = efficientnet_v2_s(
            weights="IMAGENET1K_V1",
            progress=True
        )


        # ----------------------------------------------------
        # Get feature dimension
        # EfficientNetV2-S = 1280
        # ----------------------------------------------------

        feature_channels = (
            self.backbone.classifier[1].in_features
        )


        # ----------------------------------------------------
        # Remove original ImageNet classifier
        # ----------------------------------------------------

        self.backbone.classifier = nn.Identity()


        # ----------------------------------------------------
        # CBAM
        # ----------------------------------------------------

        self.cbam = CBAM(
            channels=feature_channels
        )


        # ----------------------------------------------------
        # Global Average Pooling
        # ----------------------------------------------------

        self.avgpool = nn.AdaptiveAvgPool2d(
            (1, 1)
        )


        # ----------------------------------------------------
        # Fruit Classification Head
        # 6 classes
        # ----------------------------------------------------

        self.fruit_head = nn.Linear(
            feature_channels,
            num_fruit_classes
        )


        # ----------------------------------------------------
        # Quality Classification Head
        # 3 classes
        # ----------------------------------------------------

        self.quality_head = nn.Linear(
            feature_channels,
            num_quality_classes
        )


    def forward(self, x):

        # ----------------------------------------------------
        # EfficientNetV2-S feature extraction
        # ----------------------------------------------------

        x = self.backbone.features(x)


        # ----------------------------------------------------
        # CBAM
        # ----------------------------------------------------

        x = self.cbam(x)


        # ----------------------------------------------------
        # Global Average Pooling
        # ----------------------------------------------------

        x = self.avgpool(x)


        # ----------------------------------------------------
        # Flatten
        # ----------------------------------------------------

        x = torch.flatten(
            x,
            1
        )


        # ----------------------------------------------------
        # Fruit prediction
        # ----------------------------------------------------

        fruit_output = self.fruit_head(x)


        # ----------------------------------------------------
        # Quality prediction
        # ----------------------------------------------------

        quality_output = self.quality_head(x)


        # Return both predictions
        return fruit_output, quality_output


# ============================================================
# Create Model
# ============================================================

cbam_mtl_model = EfficientNetV2S_CBAM_MTL(
    num_fruit_classes=6,
    num_quality_classes=3
)


# ============================================================
# Device
# ============================================================

cbam_mtl_model = cbam_mtl_model.to(device)


# ============================================================
# Print Information
# ============================================================

print("CBAM + Multi-Task Learning model created successfully.")
print("=" * 60)

print("Device:", device)

print("Fruit classes:", 6)

print("Quality classes:", 3)

print("Batch size:", 8)

print("Learning rate:", 0.01)
```

```python
# ============================================================
# Cell 26 - CBAM + Multi-Task Learning Training
# ============================================================

import os
import torch
import torch.nn as nn
from tqdm.auto import tqdm

# ------------------------------------------------------------
# 🔧 Necessary modification to run the proposed MTL model
# ------------------------------------------------------------
# The original paper uses separate fruit and quality training.
# Here, both tasks are trained jointly using a combined loss.
# CutMix/MixUp is not used here because the labels must be
# mixed consistently for both fruit and quality tasks.
# ------------------------------------------------------------

# Loss functions
fruit_criterion = nn.CrossEntropyLoss()
quality_criterion = nn.CrossEntropyLoss()

# Loss weights
lambda_fruit = 1.0
lambda_quality = 1.0

# Optimizer
optimizer = torch.optim.AdamW(
    cbam_mtl_model.parameters(),
    lr=0.01,
    weight_decay=1e-2
)

# Training settings
num_epochs = 30

# Checkpoint directory
MTL_CHECKPOINT_DIR = os.path.join(
    PROJECT_PATH,
    "CBAM_MTL_Checkpoints"
)

os.makedirs(MTL_CHECKPOINT_DIR, exist_ok=True)

latest_checkpoint_path = os.path.join(
    MTL_CHECKPOINT_DIR,
    "latest_checkpoint.pth"
)

best_model_path = os.path.join(
    MTL_CHECKPOINT_DIR,
    "best_model.pth"
)

start_epoch = 0
best_val_accuracy = 0.0

# ------------------------------------------------------------
# Automatic checkpoint resume
# ------------------------------------------------------------

if os.path.exists(latest_checkpoint_path):

    checkpoint = torch.load(
        latest_checkpoint_path,
        map_location=device
    )

    cbam_mtl_model.load_state_dict(
        checkpoint["model_state_dict"]
    )

    optimizer.load_state_dict(
        checkpoint["optimizer_state_dict"]
    )

    start_epoch = checkpoint["epoch"] + 1
    best_val_accuracy = checkpoint["best_val_accuracy"]

    print("Checkpoint found.")
    print("Resuming from epoch:", start_epoch + 1)
    print("Best validation accuracy:", best_val_accuracy)

else:

    print("No previous checkpoint found.")
    print("Starting training from epoch 1.")

print("=" * 60)
print("CBAM + Multi-Task Learning Training")
print("Batch size:", 8)
print("Learning rate:", 0.01)
print("Epochs:", num_epochs)
print("Fruit loss weight:", lambda_fruit)
print("Quality loss weight:", lambda_quality)
print("=" * 60)

# ------------------------------------------------------------
# Training loop
# ------------------------------------------------------------

for epoch in range(start_epoch, num_epochs):

    cbam_mtl_model.train()

    running_loss = 0.0
    running_fruit_correct = 0
    running_quality_correct = 0
    total_samples = 0

    train_bar = tqdm(
        mtl_train_loader,
        desc=f"Epoch {epoch + 1}/{num_epochs} - Training",
        unit="batch"
    )

    for images, fruit_labels, quality_labels in train_bar:

        images = images.to(device)
        fruit_labels = fruit_labels.to(device)
        quality_labels = quality_labels.to(device)

        optimizer.zero_grad()

        fruit_outputs, quality_outputs = cbam_mtl_model(images)

        fruit_loss = fruit_criterion(
            fruit_outputs,
            fruit_labels
        )

        quality_loss = quality_criterion(
            quality_outputs,
            quality_labels
        )

        total_loss = (
            lambda_fruit * fruit_loss
            + lambda_quality * quality_loss
        )

        total_loss.backward()

        optimizer.step()

        running_loss += total_loss.item()

        fruit_predictions = torch.argmax(
            fruit_outputs,
            dim=1
        )

        quality_predictions = torch.argmax(
            quality_outputs,
            dim=1
        )

        running_fruit_correct += (
            fruit_predictions == fruit_labels
        ).sum().item()

        running_quality_correct += (
            quality_predictions == quality_labels
        ).sum().item()

        total_samples += images.size(0)

        train_bar.set_postfix(
            loss=f"{total_loss.item():.4f}",
            fruit_acc=f"{running_fruit_correct / total_samples:.4f}",
            quality_acc=f"{running_quality_correct / total_samples:.4f}"
        )

    train_loss = running_loss / len(mtl_train_loader)

    train_fruit_accuracy = (
        running_fruit_correct / total_samples
    )

    train_quality_accuracy = (
        running_quality_correct / total_samples
    )

    # --------------------------------------------------------
    # Validation
    # --------------------------------------------------------

    cbam_mtl_model.eval()

    val_running_loss = 0.0
    val_fruit_correct = 0
    val_quality_correct = 0
    val_total_samples = 0

    val_bar = tqdm(
        mtl_val_loader,
        desc=f"Epoch {epoch + 1}/{num_epochs} - Validation",
        unit="batch"
    )

    with torch.no_grad():

        for images, fruit_labels, quality_labels in val_bar:

            images = images.to(device)
            fruit_labels = fruit_labels.to(device)
            quality_labels = quality_labels.to(device)

            fruit_outputs, quality_outputs = cbam_mtl_model(images)

            fruit_loss = fruit_criterion(
                fruit_outputs,
                fruit_labels
            )

            quality_loss = quality_criterion(
                quality_outputs,
                quality_labels
            )

            total_loss = (
                lambda_fruit * fruit_loss
                + lambda_quality * quality_loss
            )

            val_running_loss += total_loss.item()

            fruit_predictions = torch.argmax(
                fruit_outputs,
                dim=1
            )

            quality_predictions = torch.argmax(
                quality_outputs,
                dim=1
            )

            val_fruit_correct += (
                fruit_predictions == fruit_labels
            ).sum().item()

            val_quality_correct += (
                quality_predictions == quality_labels
            ).sum().item()

            val_total_samples += images.size(0)

            val_bar.set_postfix(
                loss=f"{total_loss.item():.4f}"
            )

    val_loss = val_running_loss / len(mtl_val_loader)

    val_fruit_accuracy = (
        val_fruit_correct / val_total_samples
    )

    val_quality_accuracy = (
        val_quality_correct / val_total_samples
    )

    # Average validation accuracy for selecting best model
    combined_val_accuracy = (
        val_fruit_accuracy + val_quality_accuracy
    ) / 2

    # --------------------------------------------------------
    # Print epoch results
    # --------------------------------------------------------

    print()
    print(f"Epoch {epoch + 1}/{num_epochs}")
    print(f"Training Loss: {train_loss:.4f}")
    print(f"Training Fruit Accuracy: {train_fruit_accuracy:.4f}")
    print(f"Training Quality Accuracy: {train_quality_accuracy:.4f}")
    print(f"Validation Loss: {val_loss:.4f}")
    print(f"Validation Fruit Accuracy: {val_fruit_accuracy:.4f}")
    print(f"Validation Quality Accuracy: {val_quality_accuracy:.4f}")
    print(f"Combined Validation Accuracy: {combined_val_accuracy:.4f}")

    # --------------------------------------------------------
    # Save latest checkpoint
    # --------------------------------------------------------

    torch.save(
        {
            "epoch": epoch,
            "model_state_dict": cbam_mtl_model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "best_val_accuracy": best_val_accuracy
        },
        latest_checkpoint_path
    )

    print(
        "Latest checkpoint saved:",
        latest_checkpoint_path
    )

    # --------------------------------------------------------
    # Save best model
    # --------------------------------------------------------

    if combined_val_accuracy > best_val_accuracy:

        best_val_accuracy = combined_val_accuracy

        torch.save(
            cbam_mtl_model.state_dict(),
            best_model_path
        )

        print(
            f"Best model saved. "
            f"Combined validation accuracy: "
            f"{best_val_accuracy:.4f}"
        )

    print("-" * 60)

print()
print("CBAM + Multi-Task Learning Training Completed")
print("Best Combined Validation Accuracy:",
      f"{best_val_accuracy:.4f}")
print("Best Model Path:", best_model_path)
```
